```python
# Create a dictionary to store customer products
customer_products = {}
for order in customer_orders:
    customer = order[0]
    product = order[1]
    if customer in customer_products:
        customer_products[customer].append(product)
    else:
        customer_products[customer] = [product]

# Print the dictionary in a readable format
print("\nCustomer Products:")
#print(customer_products)
for customer, products in customer_products.items():
    print(f"{customer}: {', '.join(products)}")
```

```python
# Create a dictionary to store product categories
product_category = {}
for order in customer_orders:
    product = order[1]
    category = order[3]
    product_category[product] = category

# Print the dictionary in a readable format
print("\nProduct Categories:")
for product, category in product_category.items():
    print(f"{product}: {category}")
    
```

```python
# Create a set of unique product categories 
unique_categories = set(product_category.values())
# Print the unique categories with a suitable heading
print("\nAvailable Product Categories:")
for category in unique_categories:
    print(category)
print(f"\nTotal number of unique categories: {len(unique_categories)}")
```

```python
# Create a dictionary to store products by category
category_products = {}
for product, category in product_category.items():
    if category not in category_products:
        category_products[category] = []
    category_products[category].append(product)

# Print the result in a readable format
for category, products in category_products.items():
    print(f"\n{category}:")
    for product in products:
        print(f"- {product}")
        
```

```python
# Print the orders in a formatted table
print("\nCustomer Orders Table:")
print("Customer Name | Product | Price | Category ")
print("-" * 50)
for order in customer_orders:
    customer_name = order[0]
    product = order[1]
    price = order[2]
    category = order[3]
    print(f"{customer_name} | {product} | ${price:.2f} | {category}")
```

```python
# Create a dictionary to store total spending by each customer
customer_spending = {}
for order in customer_orders:
    customer_name = order[0]
    price = order[2]
    customer_spending[customer_name] = customer_spending.get(customer_name, 0) + price

# Print the total spending by each customer
print("\nCustomer Spending:")
for customer, spending in customer_spending.items():
    print(f"{customer} : ${spending:.2f}")
```

```python
# Create a dictionary to store customer classification
customer_classification = {}
for customer, spending in customer_spending.items():
    if spending > 500:
        customer_classification[customer] = "High-Value Buyer"
    elif spending >= 200:
        customer_classification[customer] = "Moderate Buyer"
    else:
        customer_classification[customer] = "Low-Value Buyer"

# Print each customer's name, total spending, and classification
print("\nCustomer Classification:")
for customer, classification in customer_classification.items():
    spending = customer_spending[customer]
    print(f"{customer} : ${spending:.2f} - {classification}")
```

```python
# Create a dictionary to count the number of customers in each classification
classification_count = {"High-Value Buyer": 0, "Moderate Buyer": 0, "Low-Value Buyer": 0}
for classification in customer_classification.values():
    classification_count[classification] += 1

# Print the results
print("\nClassification Count:")
for classification, count in classification_count.items():
    print(f"{classification}: {count}")
```

```python
# Create a dictionary to store total revenue by category
category_revenue = {}
for _, _, price, category in customer_orders:
    if category in category_revenue:
        category_revenue[category] += price
    else:
        category_revenue[category] = price

# Print the results
print("\nCategory Revenue:")
for category, revenue in category_revenue.items():
    print(f"{category} : ${revenue:.2f}")

# Identify and print the summary statistics
if category_revenue:
    print(f"\nCategory with highest revenue: {max(category_revenue, key=category_revenue.get)}")
    print(f"Category with lowest revenue: {min(category_revenue, key=category_revenue.get)}")
    print(f"Average revenue across all categories: ${sum(category_revenue.values()) / len(category_revenue):.2f}")
```

```python
# Create a set of unique product names
unique_products = set(order[1] for order in customer_orders)

# Print the results
print("\nUnique Products Sold:")
for product in sorted(unique_products):
    print(f"- {product}")
print(f"\nTotal number of unique products: {len(unique_products)}")
```

```python
# Identify the top three highest-spending customers
top_customers = sorted(customer_spending.items(), key=lambda x: x[1], reverse=True)[:3]

# Print the ranking table
print("\nTop Three Highest-Spending Customers:")
print("Rank | Customer Name | Total Spending")
print("-----|---------------|----------------")
for i, (customer, spending) in enumerate(top_customers, start=1):
    print(f"{i}    | {customer} | ${spending:.2f}")
```

```python
# Create a dictionary to count the frequency of each product
product_frequency = {}
for order in customer_orders:
    product = order[1]
    product_frequency[product] = product_frequency.get(product, 0) + 1

# Print the results
print("\nProduct Purchase Frequencies:")
for product, count in product_frequency.items():
    print(f"- {product}: {count}")

# Find the most frequently purchased product
most_frequent_product = max(product_frequency, key=product_frequency.get)
print(f"\nMost Frequently Purchased Product: {most_frequent_product}")

# Find the top three most frequently purchased products
top_three_products = sorted(product_frequency.items(), key=lambda x: x[1], reverse=True)[:3]
print("\nTop Three Most Frequently Purchased Products:")
for i, (product, count) in enumerate(top_three_products, start=1):
    print(f"{i}. {product}: {count}")
```

```python
# Create a dictionary to store categories purchased by each customer
customer_categories = {}

for order in customer_orders:
    customer = order[0]
    category = order[3]
    if customer not in customer_categories:
        customer_categories[customer] = set()
    customer_categories[customer].add(category)

# Identify customers who purchased from more than one category
multiple_category_customers = [customer for customer, categories in customer_categories.items() if len(categories) > 1]

# Print the results
print("Customers who purchased from multiple categories:")
for customer in multiple_category_customers:
    print(f"Customer: {customer}")
    print(f"Categories: {', '.join(customer_categories[customer])}")
print(f"\nTotal number of customers who purchased from multiple categories: {len(multiple_category_customers)}")
```

```python
# Create sets for electronics buyers and clothing buyers
electronics_buyers = {order[0] for order in customer_orders if order[3] == "Electronics"}
clothing_buyers = {order[0] for order in customer_orders if order[3] == "Clothing"}
common_customers = electronics_buyers & clothing_buyers

# Print the results
print("Electronics Buyers:")
for customer in electronics_buyers:
    print(f" - {customer}")

print("\nClothing Buyers:")
for customer in clothing_buyers:
    print(f" - {customer}")

print("\nCustomers who bought both Electronics and Clothing:")
for customer in common_customers:
    print(f" - {customer}")

print(f"\nTotal count of common customers: {len(common_customers)}")
```

```python
# Demonstrate set operations
union_customers = electronics_buyers | clothing_buyers
intersection_customers = electronics_buyers & clothing_buyers
difference_electronics = electronics_buyers - clothing_buyers
difference_clothing = clothing_buyers - electronics_buyers
symmetric_difference_customers = electronics_buyers ^ clothing_buyers
print("Union of Electronics and Clothing Buyers:")
for customer in union_customers:
    print(f" - {customer}")
print("\nIntersection of Electronics and Clothing Buyers:")
for customer in intersection_customers:
    print(f" - {customer}")
print("\nDifference (Electronics only):")
for customer in difference_electronics:
    print(f" - {customer}")
print("\nDifference (Clothing only):")
for customer in difference_clothing:
    print(f" - {customer}")
print("\nSymmetric Difference:")
for customer in symmetric_difference_customers:
    print(f" - {customer}")
```
